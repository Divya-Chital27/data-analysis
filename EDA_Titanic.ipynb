- the structure and type of the data
- distributions and patterns
- missing values
- unusual observations and potential outliers
- relationships between variables
- useful opportunities for feature engineering

3. Perform **Bivariate Analysis**

# 2. Univariate Analysis

**Univariate analysis** means analyzing **one variable at a time**.

The main goals are to understand the variable's distribution and identify potential data-quality or statistical issues.

### Main objectives

- **Distribution analysis:** Understand how the values are spread.
- **Identifying potential issues:** Look for missing values, unusual values, extreme observations, or unexpected patterns.

## Steps for Univariate Analysis of Numerical Columns

1. **Descriptive statistics** — summarize the variable using count, mean, standard deviation, minimum, quartiles, and maximum.
2. **Visualization** — use plots such as histograms, KDE plots, and box plots to understand the distribution visually.
3. **Identifying outliers** — investigate observations that are unusually far from the rest of the data.
4. **Skewness** — measure the asymmetry of a numerical distribution.
5. **Conclusions** — summarize the important findings and decide whether the variable needs further investigation or transformation.

### Descriptive Statistics

`describe()` provides a statistical summary of the `Age` column.

Important measures:

- **count:** number of non-missing observations
- **mean:** arithmetic average
- **std:** standard deviation, which measures spread around the mean
- **min:** minimum observed value
- **25%:** first quartile (Q1)
- **50%:** median (Q2)
- **75%:** third quartile (Q3)
- **max:** maximum observed value

For this dataset, 714 Age values are available, with a mean of about 29.70 years and a median of 28 years.


```python
df['Age'].describe()
```

### Histogram

A **histogram** divides numerical values into intervals called bins and shows how many observations fall into each interval.

Here, `bins=20` divides the Age range into 20 intervals.

A histogram helps us identify:
- the general shape of the distribution
- concentration of observations
- possible gaps
- possible extreme values


```python
df['Age'].plot(kind='hist',bins=20)
```

### KDE Plot

A **KDE (Kernel Density Estimate) plot** provides a smoothed estimate of the probability density of a numerical variable.

Unlike a histogram, which uses bins, KDE represents the distribution as a smooth curve.

It helps us visually understand:
- where observations are concentrated
- the approximate shape of the distribution
- whether the distribution is symmetric or skewed


```python
df['Age'].plot(kind='kde')
```

### Skewness

**Skewness** measures the asymmetry of a numerical distribution.

- A value close to **0** suggests a roughly symmetric distribution.
- **Positive skewness** indicates a longer/right tail.
- **Negative skewness** indicates a longer/left tail.

The value returned here helps quantify the shape observed in the plots.


```python
df['Age'].skew()
```

### Box Plot

A **box plot** summarizes a numerical distribution using:

- minimum / lower whisker
- Q1
- median
- Q3
- maximum / upper whisker
- potential outlier points

The box represents the interquartile range (IQR):

**IQR = Q3 − Q1**

Potential outliers are commonly investigated using the rule:

**Lower bound = Q1 − 1.5 × IQR**

**Upper bound = Q3 + 1.5 × IQR**

A point outside these limits is a statistical potential outlier, but it is not automatically an invalid observation.


```python
df['Age'].plot(kind='box')
```

### Investigating High Age Values

Values above 65 are inspected separately.

This is an important EDA practice: **an extreme value should be investigated before being removed**.

An elderly passenger is a perfectly valid real-world observation, so a high Age value should not be removed merely because it looks unusual in a box plot.

Therefore, these observations can be **statistical extremes without being data errors**.


```python
df[df['Age']>65]#not outliers valid data
```

### Checking Missing Values

The proportion of missing Age values is calculated as:

`number of missing values / total number of rows`

For this dataset, the result is approximately **19.9%**, which is commonly summarized as about 20%.

Missing values are important because they can affect:
- statistical summaries
- visualizations
- model training
- later feature engineering


```python
df.Age.isnull().sum()/len(df['Age'])
```

# 3. Bivariate Analysis

**Bivariate analysis** means studying the relationship between **two variables**.

While univariate analysis asks:

> “What does one variable look like?”

bivariate analysis asks:

> “How does one variable change with, or differ across, another variable?”

For categorical variables, **cross-tabulation** is useful for comparing category combinations.

In this notebook, survival is compared with:
- passenger class
- sex
- embarkation port

Other numerical relationships are explored later through distributions and feature engineering.


###steps of doing bivariate analysis


### Survival vs Passenger Class — Cross-Tabulation

`pd.crosstab()` creates a frequency table between two categorical variables.

`normalize='columns'` converts each column into proportions, and multiplying by 100 converts them into percentages.

This allows us to compare the survival percentage **within each passenger class** rather than only comparing raw counts.


```python
pd.crosstab(df['Survived'],df['Pclass'],normalize='columns')*100
```

Here, it visualizes the percentage table produced by `crosstab()` and makes differences between passenger classes easier to compare.


```python
sns.heatmap(pd.crosstab(df['Survived'],df['Pclass'],normalize='columns')*100)
```

## Creating `individual_fare`

The original Fare can represent a group booking. To estimate the fare associated with one passenger, the notebook divides Fare by:

**SibSp + Parch + 1**

The `+1` includes the passenger themself.

### Formula

`individual_fare = Fare / (SibSp + Parch + 1)`

This is a derived feature and should be interpreted as an estimated per-person fare based on the family/group count.


```python
df['individual_fare']=df['Fare']/(df['SibSp']+df['Parch']+1)
```

```python
df
```

### Checking Individual Fare with a Box Plot

A box plot is used to inspect the distribution of the newly created `individual_fare` feature and identify unusually high or low values.


```python
df['individual_fare'].plot(kind='box')
```

## Creating `Family_size`

The notebook creates:

**Family_size = SibSp + Parch + 1**

The `+1` represents the passenger themself.

This produces a more intuitive measure of the total number of family members travelling with or associated with the passenger.


```python
df['Family_size']=df['Parch']+df['SibSp']+1
```

## Creating `Family_type`

To simplify family-size analysis, passengers are grouped into three categories:

- **alone** → Family_size = 1
- **small** → Family_size from 2 to 4
- **large** → Family_size greater than 4

This converts a numerical feature into a categorical feature that can be compared more easily with survival.


```python
#family type
#1alone
#2-4 small
#5 large
def transform_family_size(num):
  if num==1:
    return 'alone'
  elif num>1 and num<5:
    return 'small'
  else:
    return 'large'

df['Family_type']=df['Family_size'].apply(transform_family_size)
```

### Survival vs Family Type

The cross-tabulation compares survival percentages across the `Family_type` categories.

This helps investigate whether travelling alone, in a small family, or in a large family was associated with different survival rates.


```python
pd.crosstab(df['Survived'],df['Family_type'],normalize='columns')*100
```

## Extracting `Surname`

The surname is extracted from the `Name` column.

Because Titanic names are stored in a format such as:

`Surname, Title. FirstName`

splitting at the comma and taking the first part gives the surname.

Surname can later be useful for identifying family groups.


```python
df['Surname']=df['Name'].str.split(',').str.get(0)
```

## Extracting Passenger `title`

The title is extracted from the `Name` field.

The process:
1. Split the name at the comma.
2. Take the second part.
3. Remove leading/trailing spaces.
4. Split the remaining text by spaces.
5. Take the first word, which contains the title.

This creates a structured `title` feature from the original text column.


```python
df['title']=df['Name'].str.split(',').str.get(1).str.strip().str.split(' ').str.get(0)
```

### Grouping Rare Titles

Several uncommon titles are grouped into the category `other`.

This reduces the number of very small categories and creates a more useful categorical feature for analysis.

Grouping rare categories is a common feature-engineering technique when individual categories have very few observations.


```python
df['title'] = df['title'].str.replace('Rev.','other')
df['title'] = df['title'].str.replace('Dr.','other')
df['title'] = df['title'].str.replace('Col.','other')
df['title'] = df['title'].str.replace('Major.','other')
df['title'] = df['title'].str.replace('Capt.','other')
df['title'] = df['title'].str.replace('the','other')
df['title'] = df['title'].str.replace('Jonkheer.','other')
# ,'Dr.','Col.','Major.','Don.','Capt.','the','Jonkheer.']
```

### Removing the Redundant Title Column

The notebook removes the old title column after the new standardized `title` feature has been created.

The final cleaned feature should use the lowercase column name `title` consistently.


### Correcting a Title Typo

The value `ootherr` is corrected to `other`.

This is a simple **data-consistency correction** caused by a spelling/typing error.


```python
df['title'] = df['title'].str.replace('ootherr','other')
```

### Checking Title Frequencies

`value_counts()` is used to verify the title categories after standardization and correction.


```python
df.title.value_counts()
```

### Selecting Common Titles

A temporary DataFrame is created containing the most relevant title categories:

- other

```python
temp_df=df[df.title.isin(['Mr.','Miss.','Mrs.','Master.','other'])]
```
